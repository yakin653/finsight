# FinSight — Phase 2 : EDA

**Objectif** : explorer les données de marché stockées dans PostgreSQL,
calculer des indicateurs et préparer les features pour le ML.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
from dotenv import load_dotenv

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)

load_dotenv()

DATABASE_URL = os.getenv("DATABASE_URL")
print("DATABASE_URL =", DATABASE_URL)
```

```python
engine = create_engine(DATABASE_URL)

with engine.connect() as conn:
    symbols = pd.read_sql("SELECT DISTINCT symbol FROM market_data ORDER BY symbol", conn)

print(symbols)
```

```python
df = pd.read_sql(
    "SELECT * FROM market_data WHERE symbol = 'AAPL' ORDER BY \"Date\"",
    engine,
    parse_dates=["Date"],
)

print(f"Dimensions : {df.shape}")
print(f"Période    : {df['Date'].min()} → {df['Date'].max()}")
df.head()
```
